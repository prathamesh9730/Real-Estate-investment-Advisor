```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df_clean['Price_Category'] = le.fit_transform(df_clean['Price_Category'])
```

```python
x = df_clean.drop(columns=['Property_ID','Price_Category'])
y = df_clean['Price_Category']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test, y_train, y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_prev = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score, f1_score, recall_score
```

```python
print("accuracy",accuracy_score(y_test, y_prev))
print("precision",precision_score(y_test,y_prev,average='weighted'))
print("F1",f1_score(y_test,y_prev,average='weighted'))
print("recall",recall_score(y_test, y_prev,average='weighted'))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
ref_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=6,
    random_state=42
)
```

```python
ref_model.fit(x_train, y_train)
```

```python
y_prev_ref =ref_model.predict(x_test)
```

```python
print("accuracy",accuracy_score(y_test, y_prev_ref))
print("precision",precision_score(y_test,y_prev_ref,average='weighted'))
print("F1",f1_score(y_test,y_prev_ref,average='weighted'))
print("recall",recall_score(y_test, y_prev_ref,average='weighted'))
```

```python
import joblib

joblib.dump(ref_model, "best_classifier.pkl")
print("✅ Best Classification Model Saved Successfully!")
```
